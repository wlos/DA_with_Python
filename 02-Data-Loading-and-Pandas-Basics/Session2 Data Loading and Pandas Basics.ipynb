## 2. Loading and Saving CSV/Excel Data

Pandas can load data from files and save tables back to files. Two common file types are CSV and Excel.

CSV and Excel comparison:

| File type | What it is | Differences |
| --- | --- | --- |
| CSV | plain text file with values separated by commas | one table per file |
| Excel | workbook file that can contain multiple worksheets | spreadsheets with multiple sheets|

### 2.1 Loading data

Pandas loading functions need a file path. The path can point to a file on your computer or to a data file on the web. In this notebook, each data file path is defined just before it is first used.

Common loading functions:

| Function | What it loads | How to use it |
| --- | --- | --- |
| `pd.read_csv(path)` | CSV files | `variable_name = pd.read_csv(DATA_PATH)` |
| `pd.read_excel(path, sheet_name=...)` | Excel files | `variable_name = pd.read_excel(DATA_PATH, sheet_name=0)` |

The `sheet_name` parameter chooses which worksheet to read from an Excel file. Use `sheet_name=0` for the first sheet, or use a sheet name such as `sheet_name='Sheet1'`.

Next, use one CSV example and one Excel example.


Use `pd.read_csv()` to read one CSV file into a variable. The result is a pandas `DataFrame`.


```python
# Define the CSV path just before using it.
BASE_PATH = 'https://raw.githubusercontent.com/wlos/DA_with_Python/main/sample_data/'
MARKETING_PATH = BASE_PATH + 'marketing_spend_wide.csv'

# Load the CSV file into a DataFrame named marketing.
marketing = pd.read_csv(MARKETING_PATH)
marketing
```

Use `pd.read_excel()` to read one Excel file into a variable. Here, `sheet_name=0` reads the first worksheet.


```python
# Define the Excel path just before using it.
TARGETS_XLSX_PATH = BASE_PATH + 'state_targets.xlsx'

# Load the first worksheet from the Excel file.
targets_from_excel = pd.read_excel(TARGETS_XLSX_PATH, sheet_name=0)
targets_from_excel
```

### 2.2 Saving data

After working with a table, we can save it to a new file. Common saving methods are:

| Method | What it saves | How to use it |
| --- | --- | --- |
| `.to_csv(path, index=False)` | CSV file | `dataframe_name.to_csv(OUTPUT_PATH, index=False)` |
| `.to_excel(path, sheet_name=..., index=False)` | Excel file | `dataframe_name.to_excel(OUTPUT_PATH, sheet_name='Sheet1', index=False)` |

Use `index=False` when you do not want to save the pandas row index as an extra column.


Save the `marketing` table as a CSV file.


```python
# Save the DataFrame without writing the pandas row index.
marketing.to_csv('marketing_saved.csv', index=False)
```

Save the same table as an Excel file. The `sheet_name` parameter names the worksheet in the new workbook.


```python
# Save the DataFrame to an Excel worksheet named marketing.
marketing.to_excel('marketing_saved.xlsx', 
                   sheet_name='marketing', 
                   index=False)
```

## 3. Series, DataFrames, and Python Data Structures

Python and pandas both use data structures to store values. A data structure is a way to organise data so we can access, transform, and analyse it.

In this notebook, we will compare four common structures: Python `list`, Python `dict`, pandas `Series`, and pandas `DataFrame`.


A `list` stores an array of items. The items can be numbers, text, Boolean values, or other Python objects. A `dict` stores paired information as key-value pairs.


```python
# A list stores an array of items.
sample_spend_list = [1200, 1350, 1500]

# A dictionary stores paired information as key-value pairs.
sample_marketing_dict = {
    'month': ['Jan', 'Feb', 'Mar'],
    'digital': [1200, 1350, 1500],
    'events': [400, 450, 500]
}

print(type(sample_spend_list))
print(type(sample_marketing_dict))
```

A list can be converted to a pandas `Series`. A `Series` is one column of data with an index.


```python
# Convert the Python list into one pandas Series.
sample_spend_series = pd.Series(sample_spend_list, name='digital')
sample_spend_series
```

A dictionary can be converted to a pandas `DataFrame`. When each dictionary value is a list, each key becomes a column name.


```python
# Convert the dictionary into a table.
sample_marketing_df = pd.DataFrame(sample_marketing_dict)
sample_marketing_df
```

To read one column from a table, use square brackets with the column name: `dataframe_name['column_name']`. The result is a `Series`.


```python
# Compare a full DataFrame with one selected Series column.
print(type(marketing))
print(type(marketing['digital']))

marketing['digital']
```

To create a new column, write the table name, then the new column name in square brackets: `dataframe_name['new_column'] = values`. The values can come from a list or from an existing column.


```python
# Create a new column from a list.
sample_marketing_df['channel'] = ['paid', 'paid', 'paid']

# Create a new column from an existing column.
sample_marketing_df['digital_copy'] = sample_marketing_df['digital']

sample_marketing_df
```

Comparison of the four data structures:

| Data structure | Comes from | Best for | Example |
| --- | --- | --- | --- |
| `list` | Python | array of items | `[1200, 1350, 1500]` |
| `dict` | Python | paired information | `{'digital': [1200, 1350, 1500]}` |
| `Series` | pandas | one table column | `marketing['digital']` |
| `DataFrame` | pandas | full table with rows and columns | `marketing` |

## 5. Indexing

Indexing means selecting part of a table. We can select one row, multiple rows, one column, or multiple columns.

Common indexing patterns:

| Goal | Pattern |
| --- | --- |
| Read one column | `df['column_name']` |
| Read multiple columns | `df[['col1', 'col2']]` |
| Read one row by position | `df.iloc[0]` |
| Read multiple rows by position | `df.iloc[[0, 1, 2]]` |
| Read multiple rows and columns by position | `df.iloc[[0, 2], [1, 3]]` |
| Use row/column labels | `df.loc[row_label, column_label]` |
| Use integer positions | `df.iloc[row_position, column_position]` |

A list inside square brackets lets us select multiple columns or multiple row positions.


Read one row by integer position with `.iloc`. The first row is position `0`.


```python
# Select the first row by integer position.
marketing.iloc[0]
```

Read multiple columns by passing a list of column names.


```python
# Select multiple columns with a list of column names.
marketing[['month', 'digital', 'events']]
```

Read multiple rows by passing a list of row positions to `.iloc`.


```python
# Select multiple rows with a list of row positions.
marketing.iloc[[0, 1, 2]]
```

Use `.loc` when selecting by row and column labels. Use `.iloc` when selecting by integer positions.


```python
# .loc uses row and column labels.
marketing.loc[0:2, ['month', 'digital']]
```

```python
# .iloc uses integer row and column positions.
marketing.iloc[0:3, 0:4]
```

Use lists inside `.iloc` to select specific rows and specific columns at the same time.


```python
# Select specific row positions and column positions with lists.
marketing.iloc[[0, 2, 4], [0, 1, 3]]
```

## 6. Common Operations

Common operations transform, compare, summarise, or reorder table values.

| Operation | Example pattern |
| --- | --- |
| Add, subtract, multiply, divide columns | `df['a'] + df['b']` |
| Compare values | `df['a'] > 100` |
| Sum values | `df['column'].sum()` |
| Find maximum value | `df['column'].max()` |
| Find minimum value | `df['column'].min()` |
| Calculate mean | `df['column'].mean()` |
| Calculate median | `df['column'].median()` |
| Sort by values | `df.sort_values('column')` |
| Sort by index | `df.sort_index()` |
| Rank values | `df['column'].rank()` |
| Count values | `df['column'].value_counts()` |


Columns can be used in arithmetic expressions. The calculation is applied row by row.


```python
# Create new calculated columns from existing numeric columns.
marketing['total_spend'] = (
    marketing['digital'] + marketing['catalogue'] + marketing['events']
)
marketing['digital_minus_events'] = marketing['digital'] - marketing['events']
marketing['digital_per_event'] = marketing['digital'] / marketing['events']
marketing['digital_times_events'] = marketing['digital'] * marketing['events']

marketing.head()
```

Comparison operations create Boolean values (`True` or `False`). 
The same Boolean result can be used as a condition to filter rows.


```python
# Store the comparison result in a new Boolean column.
marketing['high_total_spend'] = marketing['total_spend'] >= 5000

marketing[['month', 'total_spend', 'high_total_spend']]
```

Use a comparison inside square brackets to filter rows. Only rows where the comparison is `True` are kept.


```python
# Filter rows using the comparison result.
marketing[marketing['total_spend'] >= 3000]
```

Sorting and ranking help reorder or compare rows. Use `.sort_values()` to sort by column values, `.sort_index()` to sort by row index, and `.rank()` to rank values within a column.


```python
# Sort rows by total spend from largest to smallest.
marketing.sort_values('total_spend', ascending=False)
```

```python
# Sort rows by their index labels from largest to smallest.
marketing.sort_index(ascending=False)
```

```python
# Rank total spend values from highest to lowest.
marketing['total_spend'].rank(ascending=False)
```

Summary functions reduce a numeric column to one value. Common examples are `.sum()`, `.max()`, `.min()`, `.mean()`, and `.median()`.


```python
# Summarise one numeric column with common summary functions.
print('Sum:', marketing['total_spend'].sum())
print('Max:', marketing['total_spend'].max())
print('Min:', marketing['total_spend'].min())
print('Mean:', marketing['total_spend'].mean())
print('Median:', marketing['total_spend'].median())
```

Use `.value_counts()` to count how often each value appears in a column.


```python
# Count how many times each month value appears.
marketing['month'].value_counts()
```
