### Visual Data Analysis Tools Compared with Data Analysis Using Python

Many data analysis tasks can be completed with visual tools or with Python. Both approaches are useful, but they suit different situations. In this course, Python is preferred for assessment work because it supports reproducible workflows across data wrangling, visualisation, statistical analysis, clustering, and report writing.

| Comparison point | Visual data analysis tools | Data analysis with Python |

## 2. Introduction to Python

Python is a programming language made up of a series of **programming commands**. A command tells the computer to do something, such as calculate a value, store text, load a file, draw a chart, or summarise a dataset.

Python is also a **free** application software package. If you need to use Python locally on your own computer, you must first download and install Python. It can be downloaded from the official Python website: [https://www.python.org/downloads/](https://www.python.org/downloads/).

Python can be run in three common ways:

| Method | What it means | Example use |
| --- | --- | --- |
| Terminal | Type commands directly into a command-line interface | Quick checks, running installed tools, launching Jupyter |
| `.py` file | Save commands in a Python script file and run the file | Reusable scripts, automation, repeated processing |
| Notebook | Write commands in cells alongside notes and outputs | Learning, exploratory analysis, technical reports |

For this course, notebooks are the main working format because they combine code, output, visualisations, and written explanation in one place.

## 4. Basic Commands, Variables, Functions, and Outputs

Python runs commands from top to bottom. In a notebook, each code cell can contain one command or a small group of related commands.

Good early habits:

- run cells in order
- keep code cells short
- add comments when a line may not be obvious later
- check the output before moving on


### Variables and Simple Outputs

A **variable** is used to store data so it can be reused later. In data analysis, variables might store file paths, column names, settings, summary numbers, or short pieces of text for a report.

#### Variable Names

Key rules:

- use letters, numbers, and underscores only
- do not start with a number
- remember that names are case sensitive: `score`, `Score`, and `SCORE` are different
- choose names that describe the data being stored


````python
# Variable names should be valid and descriptive.
score = 85
student_score = 92
Score = 70

print('score:', score)
print('student_score:', student_score)
print('Score:', Score)
````

#### Common Variable Types

Python chooses a variable's type from the data stored in it. Common beginner types include:

| Concept | What it means | Example |
| --- | --- | --- |
| String | Text data | `'Data wrangling'` |
| Integer | Whole number | `42` |
| Float | Decimal number | `3.14` |
| Boolean | True or false value | `True` |


#### Mathematical Operations with Variables

Numeric variables can be used in mathematical operations. Python can add (`+`), subtract (`-`), multiply (`*`), and divide (`/`) values, then save the result in a new variable.


````python
# Use numeric variables in mathematical operations.
score = 8.5
maximum_score = 10
percentage = score / maximum_score * 100

print('Percentage:', percentage, 'per cent')
````

````python
# Store different types of data about a simple course activity.
activity = 'Data visualisation checkpoint'
week = 2
submitted = True

print('Activity:', activity)
print('Week:', week)
print('Submitted:', submitted)
````

### Functions

Many Python commands are used in the form of a **function**. A function is a named command that performs a task.

A function usually has these parts:

- **function name**: the command to run, such as `print`, `round`, or `len`
- **input parameters**: the data placed inside the brackets
- **output**: the result produced by the function

The general pattern is:

```python
function_name(input_parameter)
```

Some functions display an output on screen, such as `print()`. Other functions return a value that can be stored in a variable and used later, such as `round()` or `len()`.

Common beginner functions include:

| Function | What it does | Example |
| --- | --- | --- |
| `print()` | Displays output on screen | `print('Hello')` |
| `len()` | Counts the number of items or characters | `len('data')` |
| `round()` | Rounds a number | `round(85.4)` |
| `type()` | Shows the data type of a value | `type(score)` |


````python
# Functions use input parameters inside brackets.
student_name = 'Alex'
score = 85.4

print(student_name)
print(round(score))
````

````python
# Function outputs can be stored in variables and reused.
course_area = 'data visualisation'
rounded_score = round(score)
character_count = len(course_area)

print('Rounded score:', rounded_score)
print('Number of characters:', character_count)
````

### Exercise 2: Calculate a Simple Course Summary

Use Python as a tool for a familiar analysis workflow:

1. store several scores in variables
2. calculate an average score
3. print a short interpretation

This exercise practises storing data, calculating a result, and presenting a short output clearly.


````python
# Exercise 2: store assessment component scores in separate variables.
data_wrangling_score = 78
data_visualisation_score = 82
report_writing_score = 74
statistical_analysis_score = 69
clustering_score = 73

average_score = (
    data_wrangling_score
    + data_visualisation_score
    + report_writing_score
    + statistical_analysis_score
    + clustering_score
) / 5

strongest_component = 'Data visualisation'
strongest_score = data_visualisation_score

print('Average score:', round(average_score, 1), 'per cent')
print('Strongest component:', strongest_component, strongest_score, 'per cent')
````

````python
# Exercise 2: convert the summary into a short sentence.
summary_sentence = (
    'The overall result was '
    + str(round(average_score, 1))
    + ' per cent, with the strongest performance in '
    + strongest_component.lower()
    + '.'
)

print(summary_sentence)
````
