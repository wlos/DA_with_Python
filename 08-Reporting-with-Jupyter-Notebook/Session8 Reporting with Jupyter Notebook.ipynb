## 1. Reporting in a Notebook

A Jupyter notebook can combine the complete analytical workflow in one document:

| Element | Purpose in a professional report |
| --- | --- |
| Markdown | explains the question, decisions, findings, and limitations |
| Code | records the reproducible analysis |
| Tables | presents precise values and comparisons |
| Charts | reveals patterns that are easier to see visually |
| References | identifies data, methods, and external evidence |

The reader should be able to follow the reasoning without reading every line of code.


### A useful analytical sequence

> **Question → Method → Evidence → Interpretation → Action**

A professional notebook is not simply a record of everything attempted. It is a curated report in which each output supports a stated question. Keep exploratory work in a separate notebook when it would distract the report audience.


## 2. Markdown in Notebooks

Markdown cells contain formatted narrative rather than executable Python. In JupyterLab, change a cell to **Markdown** using the cell-type menu, or press `Esc` then `M`. Run the cell to render the formatting.

Use Markdown to provide hierarchy and context. A reader should be able to skim the headings and understand the report's logic.


### 2.1 Headings and document hierarchy

Use one `#` for the report title, `##` for major sections, and `###` for subsections.

```markdown
# Sales Target and Marketing Spend Plan
## Executive Summary
## Methodology
### Data Cleaning
## Analysis and Findings
```

**Professional perspective:** Heading levels communicate the relationship between ideas. Do not select a heading level merely because of its visual size.


### 2.2 Emphasis, lists, and quotations

```markdown
**Finding:** Q2 targets were higher than Q1 targets in every state.

- NSW had the highest half-year target.
- Digital received the largest share of marketing spend.

> Targets are planned values and should not be interpreted as achieved sales.
```

**Professional perspective:** Use bold text sparingly for conclusions or decision points. Use bullet lists for comparable items, not for a sequence of disconnected paragraphs.


### 2.3 Links, inline code, and code blocks

```markdown
The data was processed with [pandas](https://pandas.pydata.org/).

Use inline code for names, functions, and short expressions. Use a fenced code block when code is an example and should not be run as part of the report.

## 3. What a Good Data Analysis Report Requires

Before drafting individual sections, understand the standard the complete report must meet.

| Dimension | What a good report demonstrates |
| --- | --- |
| Purpose and audience | The report addresses a clear problem and is understandable to both technical and non-technical readers. |
| Evidence and accuracy | Claims are supported by relevant evidence, and narrative values agree with the outputs. |
| Clarity | The problem, data, findings, and limitations are explained in plain language. |
| Transparency | Sources, assumptions, definitions, and major analytical decisions are documented. |
| Responsible communication | Visuals are accurate and accessible, and sensitive information is protected. |
| Actionability | Conclusions and recommendations are specific and linked to evidence. |

These are report-level requirements. They should guide every section rather than appear only as a final checklist. Python and code support the analysis, but they are not the main focus of the report. Lead with the problem, describe the evidence and its meaning, and provide enough technical detail for reproducibility without overwhelming non-technical readers.


### The nine-part report structure

This session uses the structure from `data-analysis-report-template.ipynb`.

| Section | Main purpose | Typical evidence |
| --- | --- | --- |
| 1. Report Title | identify topic and ownership | scope, author, date |
| 2. Executive Summary | support a quick decision | headline metrics and actions |
| 3. Introduction | define the analytical question | objectives, scope, data sources |
| 4. Methodology | make the work reproducible | collection, cleaning, transformations |
| 5. Analysis and Findings | answer the questions | tables, charts, statistical results |
| 6. Discussion | interpret significance | context, implications, limitations |
| 7. Recommendations | propose feasible action | prioritised, evidence-linked actions |
| 8. References | make sources traceable | data, methods, documentation |
| 9. Appendices | preserve useful detail | extra tests, definitions, code |

The following teaching sections use notebook numbering such as **4.1**. Inside an actual report, use the report numbering shown in the first column above.


## 4. Writing Each Report Section

The examples form a coherent report titled *State Sales Targets and Marketing Spend Plan, First Half of 2025*.


### 4.1 Report Title

**Key points:**
- state the subject, population or business unit, and reporting period
- use a specific title rather than a generic label such as *Data Report*
- include author, role or team, date, and version where relevant
- apply confidentiality markings if required

**Professional example:**

> # State Sales Targets and Marketing Spend Plan, First Half of 2025  
> **Prepared by:** XXX  
> **Date:** 18 July 2026  
> **Version:** 1.0 — Internal


### 4.2 Executive Summary

**Key points:**
- write this section after completing the analysis
- state the decision context and scope in one or two sentences
- present the most decision-relevant findings with numbers
- include the principal limitation or risk
- finish with prioritised recommendations
- keep it concise and understandable without the technical detail

**Professional example:**

> This report reviews state sales targets and marketing spend for January–June 2025. The combined state target increased from 19,600 in Q1 to 21,600 in Q2, a 10.2% rise, and NSW had the highest half-year target at 10,600. Total marketing spend was $17,805, of which digital represented $9,275 or 52.1%. These figures describe plans and allocations; no achieved sales, margin, or campaign-response data is available. Management should confirm that the Q2 target increase is supported by operational capacity and establish outcome measures before evaluating marketing effectiveness.


### 4.3 Introduction

**Key points:**
- explain the business problem and why it matters
- state one or more answerable analytical objectives
- define scope: population, geography, measures, and time period
- identify data sources and analytical tools
- note important exclusions before the reader interprets the results

**Professional example:**

> The Commercial Analytics Team was asked to summarise sales targets across WA, NSW, VIC, and QLD and review the allocation of digital, catalogue, and events spending for the first half of 2025. The analysis identifies changes between Q1 and Q2, ranks state targets, and compares marketing channels. It uses three CSV files analysed with Python, pandas, and Jupyter. Achieved sales, costs, campaign exposure, and customer outcomes are outside scope; therefore, the report evaluates planning patterns rather than performance or causation.


### 4.4 Methodology

The methodology should let a suitably skilled analyst understand and reproduce the process. Describe decisions and their effects, not just function names.


#### 4.4.1 Data Collection

**Key points:**
- name the data owner and source location
- record the extraction date, coverage period, and level of detail
- describe access conditions and known source limitations
- identify join keys and expected row counts where relevant

**Professional example:**

Three CSV files from the course GitHub repository were accessed on 18 July 2026:

- **`state_targets.csv`**: 8 state-quarter records covering Q1–Q2 2025; fields include `state`, `quarter`, and `sales_target`.
- **`product_lookup.csv`**: 4 product-category records; fields include `category`, `manager`, `base_margin_pct`, and `priority`.
- **`marketing_spend_wide.csv`**: 6 monthly records covering January–June 2025; fields contain digital, catalogue, and events spend.

The report summarises sales targets and marketing spend planning.


````python
# Import the libraries and define the source files for the data example.
import pandas as pd
import matplotlib.pyplot as plt

STATE_TARGETS_URL = 'https://raw.githubusercontent.com/wlos/DA_with_Python/refs/heads/main/sample_data/state_targets.csv'
PRODUCT_LOOKUP_URL = 'https://raw.githubusercontent.com/wlos/DA_with_Python/refs/heads/main/sample_data/product_lookup.csv'
MARKETING_SPEND_URL = 'https://raw.githubusercontent.com/wlos/DA_with_Python/refs/heads/main/sample_data/marketing_spend_wide.csv'

# Load the three datasets used in the data example.
targets = pd.read_csv(STATE_TARGETS_URL)
product_lookup = pd.read_csv(PRODUCT_LOOKUP_URL)
marketing_wide = pd.read_csv(MARKETING_SPEND_URL)

print('State targets:', targets.shape)
display(targets.head())
print('Product lookup:', product_lookup.shape)
display(product_lookup.head())
print('Marketing spend:', marketing_wide.shape)
display(marketing_wide.head())
````

#### 4.4.2 Data Cleaning

**Key points:**
- profile missing values, duplicates, types, ranges, and categories
- define how each issue was treated and how many rows were affected
- preserve raw data or make every change reproducible
- avoid deleting unusual values without a business or statistical reason
- verify that currency, dates, and units are consistent

**Professional example:**

> The three datasets were checked for missing values and duplicate business keys. Sales targets and marketing spend were also checked for negative values. No rows required removal or imputation. Month was converted to a datetime value so that calendar quarters could be derived consistently.


````python
# Profile issues that would materially affect the report.
cleaning_checks = pd.DataFrame({
    'dataset': ['targets', 'product_lookup', 'marketing_wide'],
    'rows': [len(targets), len(product_lookup), len(marketing_wide)],
    'missing_values': [
        targets.isna().sum().sum(),
        product_lookup.isna().sum().sum(),
        marketing_wide.isna().sum().sum()
    ],
    'duplicate_business_keys': [
        targets.duplicated(['state', 'quarter']).sum(),
        product_lookup.duplicated(['category']).sum(),
        marketing_wide.duplicated(['month']).sum()
    ]
})
cleaning_checks
````

#### 4.4.3 Data Transformation and Feature Preparation

**Key points:**
- define each derived feature in business terms
- state grouping and aggregation rules
- confirm that denominators are valid before calculating rates
- explain joins, unmatched records, and reshaping
- retain units in column names when ambiguity is possible

**Professional example:**

> Monthly marketing data was reshaped from wide to long form, creating one row per month-channel. Month was converted to calendar quarter. Sales targets were summed by state and quarter, while marketing spend was summed by channel and quarter. Channel spend share was calculated using total first-half spend as the denominator. The product lookup was retained for contextual reporting but was not joined to the target or marketing data because no valid shared business key exists.


````python
# Reshape and aggregate the example data for reporting.
marketing_long = marketing_wide.melt(
    id_vars='month',
    value_vars=['digital', 'catalogue', 'events'],
    var_name='channel',
    value_name='spend'
)
marketing_long['quarter'] = (
    pd.to_datetime(marketing_long['month']).dt.to_period('Q').astype(str)
)

state_target_summary = (
    targets.groupby('state', as_index=False)
           .agg(
               half_year_target=('sales_target', 'sum'),
               average_quarterly_target=('sales_target', 'mean')
           )
           .sort_values('half_year_target', ascending=False)
)

quarterly_targets = targets.pivot(
    index='state', columns='quarter', values='sales_target'
)
quarterly_marketing = marketing_long.pivot_table(
    index='quarter', columns='channel', values='spend', aggfunc='sum'
)
channel_summary = (
    marketing_long.groupby('channel', as_index=False)
                  .agg(total_spend=('spend', 'sum'))
)
channel_summary['spend_share_pct'] = (
    100 * channel_summary['total_spend'] / channel_summary['total_spend'].sum()
)

display(state_target_summary)
display(quarterly_targets)
display(quarterly_marketing)
display(channel_summary.sort_values('total_spend', ascending=False))
````

### 4.5 Analysis and Findings

**Key points:**
- organise the section around analytical questions, not a list of functions
- lead with the finding, then present the supporting table or chart
- report magnitude, direction, unit, time period, and comparison base
- distinguish commercially meaningful results from noise
- label figures so they can be understood without searching the text
- include relevant negative or unexpected results

**Professional example:**

**Analytical question:** How is first-half marketing spend distributed across digital, catalogue, and events channels?

The analysis aggregates January–June 2025 spend by channel, calculates each channel's share of total spend, and presents the result in a ranked table and bar chart.


````python
# Summarise spend for the professional example question.
spend_findings = (
    channel_summary.sort_values('total_spend', ascending=False)
                   .reset_index(drop=True)
)
spend_findings
````

````python
# Present channel allocation with a report-ready chart.
ax = spend_findings.plot(
    kind='bar',
    x='channel',
    y='total_spend',
    figsize=(8, 4.5),
    color='#54A24B',
    legend=False
)
ax.set_title('Digital has the largest first-half marketing allocation')
ax.set_xlabel('Channel')
ax.set_ylabel('Marketing spend')
ax.tick_params(axis='x', rotation=0)
ax.grid(axis='y', alpha=0.25)
plt.tight_layout()
plt.show()
````

**Professional finding:**

> First-half marketing spend totalled $17,805. Digital received $9,275, representing 52.1% of total spend, compared with $4,470 for catalogue and $4,060 for events. Digital was therefore the largest planned marketing channel. However, the data contains expenditure only; without campaign-response, revenue, or margin measures, the analysis cannot determine channel effectiveness.

**Other possible analyses and findings:**

- compare the combined state sales target between Q1 and Q2
- rank states by their total first-half sales target
- calculate the Q1-to-Q2 target change for each state
- examine monthly marketing spend trends within each channel
- compare base margin percentages across product priorities
- identify managers responsible for high-priority product categories
- assess target attainment if actual sales data becomes available


### 4.6 Discussion

**Key points:**
- interpret findings in relation to the original objective
- connect results to business, policy, or operational context
- consider plausible alternative explanations
- discuss data quality, bias, uncertainty, and generalisability
- avoid repeating every value from the findings section

**Professional example:**

> Digital's 52.1% share indicates that it was the principal planned marketing channel in the first half of 2025. However, the dataset does not contain campaign response, revenue, margin, or customer outcomes. The allocation may reflect strategy, channel cost, or operational constraints; it does not demonstrate that digital was the most effective channel. Because the data covers only six months, it also cannot establish a full-year seasonal pattern.

#### Weak and strong analytical writing

| Weak statement | Stronger professional statement |
| --- | --- |
| Sales improved in Q2. | Q2 **targets** were 10.2% above Q1; achieved sales are not available. |
| Digital marketing worked best. | Digital received 52.1% of first-half spend, but no response data is available to compare channel effectiveness. |
| NSW was the best state. | NSW had the highest half-year target; the dataset does not show actual state performance. |

Strong discussion uses precise terms, preserves the difference between plans and outcomes, and avoids causal claims that the analytical design cannot support.


### 4.7 Recommendations

**Key points:**
- link every recommendation to a finding
- state the owner, action, intended outcome, and timing where possible
- prioritise actions by impact, effort, risk, or urgency
- separate evidence-supported action from further investigation
- include a measurable indicator for follow-up

**Professional example:**

1. **Sales Operations — at the end of Q2:** compare actual sales with each state target and report attainment percentage and absolute variance.
2. **Marketing Analytics — before the next allocation cycle:** add campaign outcomes and state-level spend so channel efficiency can be assessed using revenue or margin return.
3. **Commercial Planning — during target review:** confirm that the 10.2% quarter-on-quarter increase is supported by stock, staffing, and market assumptions.


### 4.8 References

**Key points:**
- cite data sources, documentation, methods, and external claims
- include owner or author, title, date, version, and access date where available
- link to a stable source and use one citation style consistently
- identify software packages and versions when reproducibility requires them
- never present an unverified external claim as analytical evidence

**Professional example:**

- DA with Python course repository (2025), `state_targets.csv`, `product_lookup.csv`, and `marketing_spend_wide.csv`, accessed 18 July 2026: <https://github.com/wlos/DA_with_Python/tree/main/sample_data>
- Project Jupyter, *Working with Markdown Cells*, accessed 18 July 2026: <https://jupyter-notebook.readthedocs.io/en/stable/examples/Notebook/Working%20With%20Markdown%20Cells.html>
- pandas development team, *pandas documentation*, accessed 18 July 2026: <https://pandas.pydata.org/docs/>


### 4.9 Appendices

**Key points:**
- include material that supports auditability but interrupts the main story
- use appendices for data dictionaries, detailed validation, extra figures, or sensitivity tests
- reference each appendix from the main report
- do not hide essential evidence or limitations in an appendix
- remove abandoned code and irrelevant exploratory output

**Professional example:**

> **Appendix A — Metric definitions**  
> `sales_target`: planned sales value for one state-quarter.  
> `spend`: marketing expenditure for one month-channel.  
> `spend_share_pct`: channel spend divided by total first-half marketing spend.  
> `q1_to_q2_change_pct`: Q2 state target minus Q1 state target, divided by the Q1 target.
